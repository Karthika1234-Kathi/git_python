```python
from os import eventfd
# try to find number is even or odd
def is_even(num):
  """
  This function returns if a given number is even or odd
  input - any valid integr
  output-odd/even
  """
  if num%2 == 0:
    return "even"
  else:
    return "odd"
```

```python
for i in range(1,11):
  x = is_even(i)
  print(x)
```
